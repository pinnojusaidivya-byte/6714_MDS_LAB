```python
import numpy as np
a1=np.array([14,2,3,4])
a2=np.array([5,6,7,8])
print("indexing")
print("element at array[0]",a1[0])
print("element at array[1]",a1[1])
print("element at array[2]",a1[2])
print("element at array[3]",a1[3])
print("element at array[3]",a1[-1])
```

```python
a3=np.array([[14,13,12],[11,10,9]])
print("indexing in 2D array")
print("element at array[0][0]",a3[0][0])
print("element at array[0][1]",a3[0][1])
print("element at array[0][2]",a3[0][2])
print("element at array[1][0]",a3[1][0])
print("element at array[1][1]",a3[1][1])
print("element at array[1][2]",a3[1][2])
```

```python
print("slicing")
print("slicing in 1D array")
print(a1[0:2])
print(a1[2:3])
print(a1[ : ])
print(a1[: : -1])
```

```python
print("slicing in 2D array")
print(a3[0:2, : ])
print(a3[1:2 ,])
print(a3[ : ,0:2])
print(a3[ : ,2:3])
print(a3[:,:])
print(a3[::-1,::-1])
```

```python
print(np.concatenate((a1,a2)))
print("stack")
a4=np.stack((a1,a2))
print( a4)
print(a4.ndim)
#print(np.stack((a1,a2)))
print(np.hstack((a1,a2)))
print(np.vstack((a1,a2)))

```

```python
print("conditional indexing")
print(a1[a1>10])
print("conditional indexing in 2d array")
print(a3[a3>10])
```

```python
print("reshaping")
print(a3.shape)
print(a3.reshape(3,2))
print(a3.reshape(6,1))
print(a3.reshape(1,6))
```
