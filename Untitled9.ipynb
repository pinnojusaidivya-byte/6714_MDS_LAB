```python
#filling the null values

df['Age'] = df['Age'].fillna(df['Age'].median())
df['Study_Hours'] = df['Study_Hours'].fillna(df['Study_Hours'].median())
df['Attendance(%)'] = df['Attendance(%)'].fillna(df['Attendance(%)'].median())
df['Test_Score'] = df['Test_Score'].fillna(df['Test_Score'].median())

# 2. Fill categorical/text columns with Mode (most frequent value)
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
df['Grade'] = df['Grade'].fillna(df['Grade'].mode()[0])

# Note: 'Student_ID' should typically NOT be filled with statistics.
# If Student_ID is missing, fill with a placeholder string or integer:
df['Student_ID'] = df['Student_ID'].fillna('UNKNOWN')

# 3. Verify all columns are filled
print("Remaining nulls per column:\n", df.isna().sum())
```

```python
import pandas as pd
import numpy as np

# 1. Insert NaN into specific cells
df.loc[0, 'Age'] = np.nan
df.loc[2, 'Test_Score'] = np.nan
df.loc[4, 'Attendance(%)'] = np.nan

# 2. Check total null count to confirm insertion
print("Total missing values inserted:", df.isna().sum().sum())
print("\nNulls per column:\n", df.isna().sum())
```

```python
num_cols = ['Test_Score']

for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"Detected {len(outliers)} outliers in {col}")
```

```python
# List of numerical columns to clean
num_cols = ['Test_Score']

# Filter out outliers row by row across all numeric columns
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Keep only rows inside the bounds
    df = df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]

# Reset row index numbers after dropping rows
df = df.reset_index(drop=True)

# Verify new dataset shape and remaining nulls
print("Cleaned dataset shape:", df.shape)
print("Remaining nulls:", df.isna().sum().sum())
```
